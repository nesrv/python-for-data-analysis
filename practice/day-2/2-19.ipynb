```python
from time import time
def test_time(fn):
    def wrapper(*args, **kwargs):
        st = time()
        res = fn(*args, **kwargs)
        dt = time() - st
        print(dt, 'sec')
        return res
      
    return wrapper   

@test_time
def fact(x):
    f = 1
    for i in range(1,x+1):
        f *= i
    return f
```
