```python
def my_sum(*args:list|tuple, **kwargs) -> int|float:
    точность = kwargs.get("точность",5)   
    s = 0
    for a in args:
        if isinstance(a, (list, tuple)):
            s += my_sum(*a, **kwargs)
        else:
            s += a 
    s = round(s, точность)
    return s
```

```python
my_sum(1, 2, [1,2], 10.5, 1/3, (1,3,4), точность=5)
```
