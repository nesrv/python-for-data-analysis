```python

def add(a,b):
    return a + b


assert add(2,3) == 5, 'тест 1 не пройден'
assert add(2,3.0) == 5
assert add(2.5,2.5) == 5
assert add(7,-2) == 5
```

```python
def apply_discount(product, discount):  
    new_price = product["цена"] * (1 - discount)
    assert product["цена"] * 0.5 <= new_price < product["цена"], "размер скидки дб < 50"
    return new_price
    

shoes = {'имя': 'Модные туфли', 'цена': 14900}
res = apply_discount(shoes, 0.45)
```
