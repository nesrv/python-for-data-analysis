```python
string='''
Сервис по поиску работы и сотрудников
HeadHunter опубликовал подборку
высокооплачиваемых вакансий в России за ноябрь 2024 года
в Москве : на первых строчках IT-архитекторы и техлиды
'''
```

```python
stop_list = "высокооплачиваемых", "техлиды", 'и', 'за', 'в'
res = list(filter(lambda x: x not in stop_list, string.split()))
res = ' '.join(res)
print (res)
```
